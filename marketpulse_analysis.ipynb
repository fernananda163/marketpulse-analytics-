# MarketPulse Analytics
### E-commerce Marketing Performance Analysis

This project analyzes digital marketing performance across four channels — Google Ads, Meta Ads, TikTok Ads, and Email — using Python.

The analysis evaluates marketing efficiency, revenue generation, conversion performance, and return on advertising spend throughout 2025.

## Project Objectives

- Analyze marketing performance across multiple channels
- Calculate key marketing KPIs
- Compare revenue and advertising efficiency
- Identify high-performing and underperforming channels
- Analyze monthly revenue trends
- Generate actionable business insights

## Key Metrics

- CTR (Click-Through Rate)
- Conversion Rate
- CPC (Cost Per Click)
- CPA (Cost Per Acquisition)
- ROAS (Return on Ad Spend)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "channel": ["Google Ads", "Meta Ads", "TikTok Ads", "Email"],
    "ad_spend": [500, 450, 300, 100],
    "revenue": [2000, 1600, 750, 900]
}

df = pd.DataFrame(data)

df
```

```python
df["ROAS"] = df["revenue"] / df["ad_spend"]

df
```

```python
df["ROAS"] = df["ROAS"].round(2)

df
```

```python
df.sort_values("ROAS", ascending=False)
```

```python
plt.bar(df["channel"], df["ROAS"])

plt.title("ROAS por canal de marketing")
plt.xlabel("Canal")
plt.ylabel("ROAS")

plt.show()
```

```python
df.describe()
```

```python
total_spend = df["ad_spend"].sum()

total_spend
```

```python
total_revenue = df["revenue"].sum()

total_revenue
```

```python
total_roas = total_revenue / total_spend

round(total_roas, 2)
```

```python
print(f"Inversión total: ${total_spend}")
print(f"Ingresos totales: ${total_revenue}")
print(f"ROAS total: {total_roas:.2f}")
```

```python
np.random.seed(42)
```

```python
dates = pd.date_range(
    start="2025-01-01",
    end="2025-12-31",
    freq="D"
)

dates
```

```python
channels = [
    "Google Ads",
    "Meta Ads",
    "TikTok Ads",
    "Email"
]

channels
```

```python
rows = []

for date in dates:
    for channel in channels:
        rows.append({
            "date": date,
            "channel": channel
        })

marketpulse_df = pd.DataFrame(rows)

marketpulse_df
```

```python
marketpulse_df["ad_spend"] = np.random.randint(
    50,
    500,
    size=len(marketpulse_df)
)

marketpulse_df.head()
```

```python
marketpulse_df["impressions"] = np.random.randint(
    1000,
    20000,
    size=len(marketpulse_df)
)

marketpulse_df.head()
```

```python
ctr = np.random.uniform(
    0.01,
    0.08,
    size=len(marketpulse_df)
)

marketpulse_df["clicks"] = (
    marketpulse_df["impressions"] * ctr
).astype(int)

marketpulse_df.head()
```

```python
conversion_rate = np.random.uniform(
    0.01,
    0.10,
    size=len(marketpulse_df)
)

marketpulse_df["conversions"] = (
    marketpulse_df["clicks"] * conversion_rate
).astype(int)

marketpulse_df.head()
```

```python
order_value = np.random.uniform(
    30,
    150,
    size=len(marketpulse_df)
)

marketpulse_df["revenue"] = (
    marketpulse_df["conversions"] * order_value
).round(2)

marketpulse_df.head()
```

```python
marketpulse_df.shape
```

```python
marketpulse_df.info()
```

```python
marketpulse_df.isnull().sum()
```

```python
marketpulse_df.describe()
```

```python
marketpulse_df["ctr"] = (
    marketpulse_df["clicks"] / marketpulse_df["impressions"]
) * 100

marketpulse_df["ctr"] = marketpulse_df["ctr"].round(2)

marketpulse_df.head()
```

```python
marketpulse_df["conversion_rate"] = (
    marketpulse_df["conversions"] / marketpulse_df["clicks"]
) * 100

marketpulse_df["conversion_rate"] = marketpulse_df["conversion_rate"].round(2)

marketpulse_df.head()
```

```python
marketpulse_df["cpc"] = (
    marketpulse_df["ad_spend"] / marketpulse_df["clicks"]
)

marketpulse_df["cpc"] = marketpulse_df["cpc"].round(2)

marketpulse_df.head()
```

```python
marketpulse_df["cpc"] = (
    marketpulse_df["ad_spend"] / marketpulse_df["clicks"]
)

marketpulse_df["cpc"] = marketpulse_df["cpc"].round(2)

marketpulse_df.head()
```

```python
marketpulse_df["cpa"] = (
    marketpulse_df["ad_spend"] / marketpulse_df["conversions"]
)

marketpulse_df["cpa"] = marketpulse_df["cpa"].round(2)

marketpulse_df.head()
```

```python
marketpulse_df["roas"] = (
    marketpulse_df["revenue"] / marketpulse_df["ad_spend"]
)

marketpulse_df["roas"] = marketpulse_df["roas"].round(2)

marketpulse_df.head()
```

```python
channel_summary = marketpulse_df.groupby("channel").agg({
    "ad_spend": "sum",
    "revenue": "sum",
    "clicks": "sum",
    "conversions": "sum"
}).round(2)

channel_summary
```

```python
channel_summary["roas"] = (
    channel_summary["revenue"] / channel_summary["ad_spend"]
)

channel_summary["roas"] = channel_summary["roas"].round(2)

channel_summary.sort_values("roas", ascending=False)
```

```python
channel_summary["cpa"] = (
    channel_summary["ad_spend"] / channel_summary["conversions"]
)

channel_summary["cpa"] = channel_summary["cpa"].round(2)

channel_summary.sort_values("roas", ascending=False)
```

```python
import matplotlib.pyplot as plt

roas_sorted = channel_summary.sort_values("roas", ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(roas_sorted.index, roas_sorted["roas"])

plt.title("ROAS by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("ROAS")

plt.tight_layout()
plt.show()
```

```python
revenue_sorted = channel_summary.sort_values("revenue", ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(revenue_sorted.index, revenue_sorted["revenue"])

plt.title("Total Revenue by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("Revenue ($)")

plt.tight_layout()
plt.show()
```

```python
cpa_sorted = channel_summary.sort_values("cpa", ascending=True)

plt.figure(figsize=(8, 5))
plt.bar(cpa_sorted.index, cpa_sorted["cpa"])

plt.title("CPA by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("CPA ($)")

plt.tight_layout()
plt.show()
```

```python
marketpulse_df["month"] = marketpulse_df["date"].dt.to_period("M")

monthly_revenue = (
    marketpulse_df.groupby("month")["revenue"]
    .sum()
    .reset_index()
)

monthly_revenue
```

```python
monthly_revenue["month"] = monthly_revenue["month"].astype(str)

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_revenue["month"],
    monthly_revenue["revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend - 2025")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_channel = (
    marketpulse_df.groupby(["month", "channel"])["revenue"]
    .sum()
    .reset_index()
)

monthly_channel["month"] = monthly_channel["month"].astype(str)

monthly_channel.head()
```

```python
plt.figure(figsize=(11, 6))

for channel in monthly_channel["channel"].unique():
    channel_data = monthly_channel[
        monthly_channel["channel"] == channel
    ]

    plt.plot(
        channel_data["month"],
        channel_data["revenue"],
        marker="o",
        label=channel
    )

plt.title("Monthly Revenue by Marketing Channel - 2025")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=45)
plt.legend(title="Channel")

plt.tight_layout()
plt.show()
```

## Business Insights & Conclusions

The MarketPulse analysis evaluated marketing performance across Email, Google Ads, Meta Ads, and TikTok Ads during 2025.

### Key Findings

- Email generated the highest total revenue at $915,342.24.
- Email also achieved the highest ROAS at 9.10, generating approximately $9.10 in revenue for every $1 spent.
- Google Ads achieved a ROAS of 8.86 and the lowest overall CPA at approximately $10.14.
- TikTok Ads generated a ROAS of 8.01 with a CPA of approximately $11.01.
- Meta Ads had the highest total ad spend ($103,568) but the lowest ROAS (7.83) and highest CPA ($11.47) among the four channels.
- Monthly revenue peaked in May 2025 at $339,638.21.
- April 2025 recorded the lowest monthly revenue at $236,564.58.

### Business Recommendation

Based on the 2025 performance data, Email and Google Ads showed strong efficiency across ROAS and CPA. Meta Ads may warrant further campaign-level analysis to identify opportunities to improve conversion efficiency and return on advertising spend.
